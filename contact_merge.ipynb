```python
import pandas as pd
```

```python
bd_weather = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "tem": [35, 32, 30],
    "humidity": [80, 60, 78],
})

bd_weather
```

```python
us_weather = pd.DataFrame({
    "city": ["new york", "florida", "chicago"],
    "tem": [25, 22, 20],
    "humidity": [60, 40, 58],
})

us_weather
```

```python
# vertical merge
df = pd.concat([bd_weather, us_weather])
df
```

```python
df = pd.concat([bd_weather, us_weather], ignore_index=True)
df
```

```python
df = pd.concat([bd_weather, us_weather], keys=["bangladesh", "us"])
df
```

```python
# get the us cities
df.loc["us"]
```

```python
tem_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "tem": [35, 32, 30],
})

wind_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "wind": [55, 72, 90],
})

print(tem_df)
```

```python
pd.concat([tem_df, wind_df], axis=1)
```

```python
# Merge
tem_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "tem": [35, 32, 30],
})

wind_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "wind": [55, 72, 90],
})

print(wind_df)
```

```python
pd.merge(tem_df, wind_df, on="city")
```

```python
# Merge
tem_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "chittagong", "rajshahi"],
    "tem": [35, 32, 30, 45],
})

wind_df = pd.DataFrame({
    "city": ["dhaka", "khulna", "sylhet"],
    "wind": [55, 72, 90],
})

print(wind_df)
```

```python
pd.merge(tem_df, wind_df, on="city", how='outer') # how=inner/right/left/outer
```
