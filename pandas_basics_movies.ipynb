```python
# Add a new column of how old the movie is from 2026
df["age"] = df["release_year"].apply(lambda x: 2026 - x)
df.sample(5)
```

```python
# add new column of profit, revenue - budget
# 1. Drop existing 'profit' column if present
if "profit" in df.columns:
    df.drop(columns=["profit"], inplace=True)

# 2. Calculate profit
profit_data = df["revenue"] - df["budget"]

# 3. Get location of 'unit'
unit_idx = df.columns.get_loc("unit")

# 4. Insert 'profit' before 'unit'
df.insert(unit_idx, "profit", profit_data)

df.sample(5)
```

```python
# set title as index
df.set_index("title", inplace=True)
```
