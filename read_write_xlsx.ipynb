```python
dfs = pd.read_excel("movies_db.xlsx", sheet_name=["actors", "movie_actor", "languages"])
df_actor = dfs["actors"]
df_movie_actor = dfs["movie_actor"]
df_languages = dfs["languages"]
df_financials.head(5)
```

```python
# convert $$, Dollars to USD
def convert_curr(curr):
    if curr == "$$" or curr == "Dollars":
        return "USD"
    return curr

df_financials = pd.read_excel("movies_db.xlsx", "financials", converters = {
    "currency": convert_curr
})
df_financials.head(10)
```

# Merge Movies and Financials report based on movie_id

```python
df_merged = pd.merge(df_movies, df_financials, on="movie_id")
df_merged.head(2)
```

```python
# Export to new xlxs file
df_merged.to_excel("movie_finance_merged.xlsx")
```
